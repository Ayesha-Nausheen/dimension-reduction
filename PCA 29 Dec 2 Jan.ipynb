- Dimension Reduction : Reducing the number of input features
- Techniques:
    - Feature Selection : Identify unimportant variables and drop them
        - Filter Method: done before modelling
            - eg: unique(name), low correlation, low varience, no info gain and proportion)
        - Wrapper Method 
            - Forward features elimination 
            - Backwards features elimination 
            - Recursive features elimination 
        - Embedded Method : done after modelling
            - eg: Lasso regression
    - Feature Extraction: Extract important features from data

```python
X=df.iloc[:,:13]
y=df.iloc[:,13]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

# Without PCA

```python
from sklearn.linear_model import LogisticRegression
base_model=LogisticRegression()
base_model.fit(X_train,y_train)

y_pred=base_model.predict(X_test)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

# Applying PCA

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X=sc.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.decomposition import PCA
pca_model=PCA(n_components=0.95)

X_train_pca=pca_model.fit_transform(X_train)
X_test_pca=pca_model.transform(X_test)
```

```python
X_train_pca.shape
```

```python
pca_model.explained_variance_ratio_
```

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression()
classifier.fit(X_train_pca,y_train)

y_pred=classifier.predict(X_test_pca)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
